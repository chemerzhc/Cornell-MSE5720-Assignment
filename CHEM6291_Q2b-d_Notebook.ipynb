# Question 2(b–d)

- 2(b) Langmuir isotherm: θ vs E for K0 = 1, 5, 25 (E range) and θ vs log10(C) for E in [0, 0.4] V at K0=1.
- 2(c) Modified Langmuir model for O2 occupying 1 or 2 sites (θ vs E).
- 2(d) Frumkin isotherm with f = -20, -10, 0, 10, 20 (K0=1, C=1).

All plotted with T = 298.15 K; constants R and F included.


```python
import numpy as np
import matplotlib.pyplot as plt
R = 8.314462618
F = 96485.33212
T = 298.15
plt.rcParams['figure.dpi'] = 120
```

## 2(b) Langmuir isotherm

Equation:

$$\frac{\theta}{1-\theta} = K_0 C_{bulk} e^{FE/RT},\qquad \theta=\frac{K_0 C e^{FE/RT}}{1+K_0 C e^{FE/RT}}.$$

```python
E = np.linspace(-0.5, 0.8, 400)
C = 1.0
K_list = [1.0, 5.0, 25.0]
plt.figure(figsize=(6,4))
for K in K_list:
    num = K * C * np.exp(F*E/(R*T))
    theta = num / (1 + num)
    plt.plot(E, theta, label=f'K0={K}')
plt.xlabel('Potential E (V)')
plt.ylabel('Coverage θ')
plt.title('Langmuir isotherm: θ vs E (C=1)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### θ vs log10(C) for K0 = 1 at potentials E = 0, 0.1, 0.2, 0.3, 0.4 V

```python
C_vals = np.logspace(-6, 0, 400)
E_list = [0.0, 0.1, 0.2, 0.3, 0.4]
K0 = 1.0
plt.figure(figsize=(6,4))
for E in E_list:
    theta_C = (K0 * C_vals * np.exp(F*E/(R*T))) / (1 + K0 * C_vals * np.exp(F*E/(R*T)))
    plt.semilogx(C_vals, theta_C, label=f'E={E:.1f} V')
plt.xlabel('Bulk conc. C (M)')
plt.ylabel('Coverage θ')
plt.title('Langmuir: θ vs log10(C) for different potentials (K0=1)')
plt.legend()
plt.grid(alpha=0.3, which='both')
plt.show()
```

## 2(c) Modified Langmuir for O$_2$ occupying 1 or 2 adsorption sites

Case definitions and result: for m=2 the effective factor becomes 2 in the numerator under the simple assumption θ=2x.

```python
E = np.linspace(-0.5, 0.8, 400)
K = 1.0
C = 1.0
num1 = K * C * np.exp(F*E/(R*T))
theta1 = num1 / (1 + num1)
num2 = 2 * K * C * np.exp(F*E/(R*T))
theta2 = num2 / (1 + num2)
plt.figure(figsize=(6,4))
plt.plot(E, theta1, label='O2 occupies 1 site (m=1)')
plt.plot(E, theta2, label='O2 occupies 2 sites (m=2)')
plt.xlabel('Potential E (V)')
plt.ylabel('Coverage θ')
plt.title('Modified Langmuir: 1-site vs 2-site occupancy (K=1, C=1)')
plt.legend()
plt.grid(alpha=0.3)
plt.ylim(-0.05,1.05)
plt.show()
```

## 2(d) Frumkin isotherm

Equation:

$$\frac{\theta}{1-\theta} e^{f\theta} = K_0 C e^{FE/RT}$$

Solve numerically for θ in (0,1).

```python
def frumkin_theta(rhs, f, tol=1e-8, maxiter=200):
    a = 1e-12
    b = 1 - 1e-12
    def g(theta):
        return (theta/(1-theta))*np.exp(f*theta) - rhs
    ga = g(a)
    gb = g(b)
    if np.isnan(ga) or np.isnan(gb):
        return np.nan
    if ga*gb > 0:
        if ga > 0:
            return a
        else:
            return b
    for i in range(maxiter):
        m = 0.5*(a+b)
        gm = g(m)
        if abs(gm) < tol:
            return m
        if ga*gm <= 0:
            b = m
            gb = gm
        else:
            a = m
            ga = gm
    return 0.5*(a+b)

E = np.linspace(-0.6, 0.6, 400)
K0 = 1.0
C = 1.0
f_list = [-20, -10, 0, 10, 20]
plt.figure(figsize=(6,4))
for f in f_list:
    theta_vals = []
    for E_val in E:
        rhs = K0 * C * np.exp(F*E_val/(R*T))
        th = frumkin_theta(rhs, f)
        theta_vals.append(th)
    plt.plot(E, theta_vals, label=f'f={f}')
plt.xlabel('Potential E (V)')
plt.ylabel('Coverage θ')
plt.title('Frumkin isotherm: θ vs E for various f (K0=1, C=1)')
plt.legend()
plt.grid(alpha=0.3)
plt.ylim(-0.05,1.05)
plt.show()
```
